```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Step 1: Take data from user
n = int(input("How many data points? "))

X = []
y = []

print("Enter values for X and Y:")
for i in range(n):
    x_val = float(input(f"X[{i}]: "))
    y_val = float(input(f"Y[{i}]: "))
    X.append([x_val])
    y.append([y_val])

X = np.array(X)
y = np.array(y)
```

```python
# Step 2: Train model
model = LinearRegression()
model.fit(X, y)
```

```python
# Step 3: Show learned line
w = model.coef_[0][0]
b = model.intercept_[0]

print(f"\nLearned equation: y = {w:.2}x + {b:.2}")
```

```python
# Step 4: Predict new value
user_input = float(input("\nEnter X value to predict: "))
predicted_y = model.predict([[user_input]])

print("Predicted Y:", predicted_y[0][0])
```

```python
# Step 5: Plot data and line
plt.scatter(X, y, label="Data Points" , color="blue")

plt.plot(X, model.predict(X), color="purple", label="Regression Line")

plt.scatter(user_input, predicted_y[0][0],
            marker=".", s=100, label="Prediction")

plt.legend()
plt.show()
```
