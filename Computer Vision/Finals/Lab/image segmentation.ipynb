
*  cv2 → OpenCV library (used for image processing)
*   numpy → used for arrays (images are stored as arrays)
*   matplotlib.pyplot → used to display images (like graphs)


```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```



*   Import the file-handling feature from Google Colab
*   files.upload() → opens a file upload button
*   List item





```python
from google.colab import files
uploaded = files.upload()
```

*   cv2.imread() → loads the image from the given path
*   list(uploaded.keys())[0] gets the first uploaded filename.
*  Image is read in BGR format (not RGB)





```python
image = cv2.imread(list(uploaded.keys())[0])
```



*   Convert the image from BGR format to RGB format




```python
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

*   imshow() → shows image
cmap='gray' → display in grayscale
*  title() → adds title
*   axis('off') → removes x-y axis
*  show() → displays output










```python
gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis('off')
plt.show()
```

*   cv2.threshold() → converts image into binary (black & white)
*   127 → threshold value
*   pixel < 127 → black
*  pixel ≥ 127 → white
*   255 → max value (white)
*   THRESH_BINARY → type of thresholding
*   ret = threshold value used
*   thresh = resulting segmented image
*   Displays the segmented (binary) image
*   Now objects are separated from background



```python
ret, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_TOZERO)
plt.imshow(thresh, cmap='gray')
plt.title("Threshold (Segmented Image)")
plt.axis('off')
plt.show()
```

*   Creates a figure window
*   Size = width 12, height 4






```python
plt.figure(figsize=(12,4))
```

*   subplot(1,3,1) → 1 row, 3 columns, position 1

```python

plt.title("Original Image")
plt.imshow(image)
plt.axis('off')
```

*   Shows grayscale version

```python

plt.title("Grayscale Image")
plt.imshow(gray, cmap='gray')
plt.axis('off')
```

*   Shows final segmented image

```python

plt.title("Segmented Image")
plt.imshow(thresh, cmap='gray')
plt.axis('off')
```
