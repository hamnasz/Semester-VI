*   cv2 → image processing
*   matplotlib → display images
*   numpy → arrays (image = matrix)

```python
import cv2
import numpy as np
from google.colab import files
import matplotlib.pyplot as plt
```

*   Opens upload button
*   User selects image
*   Stored in uploaded dictionary


```python
uploaded = files.upload()

# Get file name
image_path = list(uploaded.keys())[0]
```

*   Reads uploaded image
*   Converts BGR → RGB

```python
img = cv2.imread(image_path)

# Convert BGR → RGB (important for correct display)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

*   Applies Mean Shift algorithm on image
*   Must be a color image (RGB or BGR)
*   sp20=Looks at neighboring pixels within radius 20
*   sr=Controls color similarity threshold

```python
# Mean Shift Filtering
# (sp = spatial radius, sr = color radius)
segmented = cv2.pyrMeanShiftFiltering(img, sp=20, sr=30)
```

*   plt.imshow() → displays an image
*   segmented → the output image from Mean Shift
*   Removes x-axis and y-axis
*   plt.show()=displays everything on screen


```python
plt.imshow(segmented)
plt.title("Mean Shift Segmented Image")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(segmented)
plt.title("Mean Shift")
plt.axis("off")

plt.show()
```
