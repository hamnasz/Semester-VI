```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

# 1. Define the full list of URLs provided
# (Grouped by Category for better data labeling)
url_map = {
    "Computing & IT": [
        "https://tuf.edu.pk/n/course-detail/bs-computer-sciences",
        "https://tuf.edu.pk/n/course-detail/bs-software-engineering",
        "https://tuf.edu.pk/n/course-detail/bs-information-technology",
        "https://tuf.edu.pk/n/course-detail/bs-artificial-intelligence",
        "https://tuf.edu.pk/n/course-detail/bs-lnternet-of-things",
        "https://tuf.edu.pk/n/course-detail/adp-computer-science",
        "https://tuf.edu.pk/n/course-detail/ms-computer-science",
        "https://tuf.edu.pk/n/course-detail/ms-computing-and-data-sciences",
        "https://tuf.edu.pk/n/course-detail/phd-computer-science-after-msmphil"
    ],
    "Medical & Allied Health": [
        "https://tuf.edu.pk/n/course-detail/doctor-of-physical-therapy",
        "https://tuf.edu.pk/n/course-detail/doctor-of-pharmacy",
        "https://tuf.edu.pk/n/course-detail/bs-optometry",
        "https://tuf.edu.pk/n/course-detail/bs-imaging-sciences",
        "https://tuf.edu.pk/n/course-detail/bs-medical-laboratory-sciences",
        "https://tuf.edu.pk/n/course-detail/bs-dietetics-and-nutrition-sciences",
        "https://tuf.edu.pk/n/course-detail/bs-anesthesia",
        "https://tuf.edu.pk/n/course-detail/bs-dental-sciences",
        "https://tuf.edu.pk/n/course-detail/bs-emergency-intensive-care-sciences",
        "https://tuf.edu.pk/n/course-detail/bs-nursing",
        "https://tuf.edu.pk/n/course-detail/bs-nursing-post-rn-bsn",
        "https://tuf.edu.pk/n/course-detail/bs-psychology"
    ],
    "Engineering": [
        "https://tuf.edu.pk/n/course-detail/bachelor-of-electrical-engineering",
        "https://tuf.edu.pk/n/course-detail/bs-civil-engineering",
        "https://tuf.edu.pk/n/course-detail/ms-electrical-engineering",
        "https://tuf.edu.pk/n/course-detail/ms-civil-engineering"
    ],
    "Management & Business": [
        "https://tuf.edu.pk/n/course-detail/bba",
        "https://tuf.edu.pk/n/course-detail/mba",
        "https://tuf.edu.pk/n/course-detail/mba-executive",
        "https://tuf.edu.pk/n/course-detail/bs-accounting-and-finance",
        "https://tuf.edu.pk/n/course-detail/bs-aviation-management",
        "https://tuf.edu.pk/n/course-detail/adp-business-administration",
        "https://tuf.edu.pk/n/course-detail/adp-accounting-and-finance"
    ],
    "Natural Sciences": [
        "https://tuf.edu.pk/n/course-detail/bs-mathematics",
        "https://tuf.edu.pk/n/course-detail/bs-physics",
        "https://tuf.edu.pk/n/course-detail/bs-biochemistry",
        "https://tuf.edu.pk/n/course-detail/bs-biotechnology"
    ],
    "Social Sciences & Humanities": [
        "https://tuf.edu.pk/n/course-detail/bachelor-of-laws",
        "https://tuf.edu.pk/n/course-detail/bs-criminology",
        "https://tuf.edu.pk/n/course-detail/bs-lnternational-relations",
        "https://tuf.edu.pk/n/course-detail/bs-mass-communication",
        "https://tuf.edu.pk/n/course-detail/bs-english-with-specialization-in-language-and-literature-translation-and-lnterpreting",
        "https://tuf.edu.pk/n/course-detail/bs-islamic-sciences",
        "https://tuf.edu.pk/n/course-detail/bs-shariah-islamic-fiqh",
        "https://tuf.edu.pk/n/course-detail/bed-hons-education",
        "https://tuf.edu.pk/n/course-detail/bed"
    ],
    "Art & Design": [
        "https://tuf.edu.pk/n/course-detail/bachelor-of-interior-design",
        "https://tuf.edu.pk/n/course-detail/bachelor-of-graphic-design"
    ],
    "Advanced Studies (MPhil/PhD)": [
        "https://tuf.edu.pk/n/course-detail/ms-physical-therapy",
        "https://tuf.edu.pk/n/course-detail/mphil-optometry",
        "https://tuf.edu.pk/n/course-detail/mphil-dietetics-nutritional-sciences",
        "https://tuf.edu.pk/n/course-detail/mphil-biotechnology",
        "https://tuf.edu.pk/n/course-detail/mphil-mathematics",
        "https://tuf.edu.pk/n/course-detail/mphil-biochemistry",
        "https://tuf.edu.pk/n/course-detail/mphil-physiology",
        "https://tuf.edu.pk/n/course-detail/mphil-pharmacology",
        "https://tuf.edu.pk/n/course-detail/mphil-pharmaceutics",
        "https://tuf.edu.pk/n/course-detail/mphil-education",
        "https://tuf.edu.pk/n/course-detail/mphil-english-literature",
        "https://tuf.edu.pk/n/course-detail/mphil-lslamic-studies",
        "https://tuf.edu.pk/n/course-detail/mphil-morbid-anatomy-and-histopathology",
        "https://tuf.edu.pk/n/course-detail/ms-management-sciences",
        "https://tuf.edu.pk/n/course-detail/ms-accounting-and-finance",
        "https://tuf.edu.pk/n/course-detail/ms-risk-and-safety-management",
        "https://tuf.edu.pk/n/course-detail/ms-physics",
        "https://tuf.edu.pk/n/course-detail/phd-dietetics-nutritional-sciences-after-msmphil",
        "https://tuf.edu.pk/n/course-detail/phd-biotechnology",
        "https://tuf.edu.pk/n/course-detail/phd-civil-engineering-after-msmphil",
        "https://tuf.edu.pk/n/course-detail/phd-electrical-engineering-after-msmphil",
        "https://tuf.edu.pk/n/course-detail/phd-mathematics-after-msmphil",
        "https://tuf.edu.pk/n/course-detail/phd-physics",
        "https://tuf.edu.pk/n/course-detail/phd-biochemistry-after-msmphil",
        "https://tuf.edu.pk/n/course-detail/phd-management-sciences-after-msmphil",
        "https://tuf.edu.pk/n/course-detail/phd-education",
        "https://tuf.edu.pk/n/course-detail/phd-islamic-studies-after-msmphil",
        "https://tuf.edu.pk/n/course-detail/phd-english-literature-after-msmphil",
        "https://tuf.edu.pk/n/course-detail/phd-pharmaceutics-after-msmphil",
        "https://tuf.edu.pk/n/course-detail/phd-pharmacology-after-msmphil"
    ]
}

# 2. Function to scrape a single course page
def scrape_course_page(category, url):
    print(f"Scraping: {url}")
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }
    
    try:
        response = requests.get(url, headers=headers)
        response.raise_for_status()
        soup = BeautifulSoup(response.content, 'html.parser')
        
        # Extract Program Name (Assuming H1 usually holds the title)
        program_name = soup.find('h1').get_text(strip=True) if soup.find('h1') else "Unknown Program"
        
        extracted_data = []
        
        # TUF usually displays semesters in tables.
        # We look for all tables and try to identify the semester from the preceding header.
        tables = soup.find_all('table')
        
        for index, table in enumerate(tables):
            # Attempt to read table using pandas for easy parsing
            try:
                # Read HTML table into dataframe
                df = pd.read_html(str(table))[0]
                
                # Basic cleaning of columns
                df.columns = [c.strip() for c in df.columns]
                
                # Check if this looks like a course list (needs specific columns)
                # Flexible check: does it have "Code" or "Title" or "Credit"?
                if any(x in str(df.columns) for x in ['Code', 'Title', 'Credit']):
                    
                    # Try to find the semester label (often in an H3 or strong tag before the table)
                    # We look at the Previous Sibling of the table in the HTML
                    semester_label = f"Semester {index + 1}" # Default if header not found
                    prev_elem = table.find_previous(['h2', 'h3', 'h4', 'strong', 'b'])
                    if prev_elem:
                        text = prev_elem.get_text(strip=True)
                        if "Semester" in text or "Year" in text:
                            semester_label = text

                    # Iterate through rows and add to list
                    for _, row in df.iterrows():
                        course_entry = {
                            "Domain": category,
                            "Program_Name": program_name,
                            "Semester": semester_label,
                            "Course_Code": row.get('Course Code', row.get('Code', '')),
                            "Course_Title": row.get('Course Title', row.get('Title', '')),
                            "Credit_Hours": row.get('Credit Hours', row.get('Cr. Hrs', '')),
                            "Category": row.get('Category', ''),
                            "Pre_Requisite": row.get('Pre-Requisite', row.get('Prerequisite', 'None')),
                            "Source_URL": url
                        }
                        extracted_data.append(course_entry)
            except Exception as e:
                print(f"Skipping a table in {url}: {e}")
                continue
                
        return extracted_data

    except Exception as e:
        print(f"Failed to scrape {url}: {e}")
        return []

# 3. Main Execution
all_course_data = []

for category, links in url_map.items():
    for link in links:
        data = scrape_course_page(category, link)
        all_course_data.extend(data)
        time.sleep(1) # Be polite to the server

# 4. Save to CSV
if all_course_data:
    df_final = pd.DataFrame(all_course_data)
    
    # Cleaning: Remove rows where Course Title is empty (usually table headers repeated)
    df_final = df_final[df_final['Course_Title'] != '']
    
    filename = "tuf_university_dataset.csv"
    df_final.to_csv(filename, index=False)
    print(f"Success! Dataset saved as {filename}")
    print(df_final.head())
else:
    print("No data extracted. Check internet connection or website structure changes.")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import re
from io import StringIO
import time

# --- CONFIGURATION ---
fee_urls = {
    "General Fee Structure": "https://tuf.edu.pk/fee-structure",
    "International Student Fees": "https://tuf.edu.pk/n/international-student"
}

policy_urls = {
    "Fee Policy": "https://tuf.edu.pk/n/tuf-fee-policy",
    "Scholarship Policy": "https://tuf.edu.pk/scholarship-policy",
    "Admission Policy": "https://tuf.edu.pk/admission-policy",
    "Rules & Regulations": "https://tuf.edu.pk/n/rules-and-regulation"
}

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

# --- HELPER: Cleaning functions ---
def clean_currency(value):
    """Converts 'Rs. 50,000' -> 50000.0"""
    if pd.isna(value):
        return None
    clean_val = re.sub(r'[^\d.]', '', str(value))
    try:
        return float(clean_val) if clean_val else None
    except ValueError:
        return None

def clean_header(col_name):
    """Flattens and cleans column names"""
    return str(col_name).strip().replace('\n', ' ').replace('  ', ' ')

# --- FUNCTION 1: Robust Fee Scraper ---
def scrape_fees_v2(url_map):
    all_fees = []
    
    for category, url in url_map.items():
        print(f"--- Scraping: {category} ---")
        try:
            response = requests.get(url, headers=headers)
            soup = BeautifulSoup(response.content, 'html.parser')
            tables = soup.find_all('table')
            print(f"Found {len(tables)} tables.")

            for i, table in enumerate(tables):
                try:
                    # 1. Parse Table (Try Multi-header first, then single)
                    html_io = StringIO(str(table))
                    try:
                        df = pd.read_html(html_io, header=[0, 1])[0]
                    except:
                        html_io.seek(0)
                        df = pd.read_html(html_io, header=0)[0]

                    # 2. Flatten Multi-Level Headers (The key fix)
                    # If columns are a MultiIndex, we want to extract the Faculty Name
                    faculty_name = "General"
                    if isinstance(df.columns, pd.MultiIndex):
                        # Join the levels to check for keywords
                        flat_cols = []
                        for col in df.columns:
                            # If it's a tuple like ('Serial', 'Faculty of Science'), grab 'Faculty of Science'
                            # We look for the longest string in the tuple that isn't the column name itself
                            parts = [str(p) for p in col if "Unnamed" not in str(p)]
                            flat_cols.append(" ".join(parts).strip())
                            
                            # Try to find the Faculty name from the header
                            for p in parts:
                                if "Faculty" in p or "Department" in p:
                                    faculty_name = p
                        
                        df.columns = flat_cols
                    else:
                        df.columns = [clean_header(c) for c in df.columns]

                    # 3. Validation: Does this look like a fee table?
                    # We check the flattened columns for keywords
                    col_str = " ".join(df.columns).lower()
                    keywords = ['program', 'degree', 'fee', 'tuition', 'amount', 'semester']
                    
                    if any(k in col_str for k in keywords):
                        print(f"  [Table {i}] Accepted. (Faculty detected: {faculty_name})")
                        
                        # Add Metadata
                        df['Category'] = category
                        df['Faculty'] = faculty_name
                        df['Source_URL'] = url
                        
                        # Standardize Columns
                        # We try to rename columns to a clean format for ML
                        rename_map = {}
                        for col in df.columns:
                            cl = col.lower()
                            if 'discipline' in cl or 'program' in cl:
                                rename_map[col] = 'Program'
                            elif 'total' in cl and 'fee' in cl:
                                rename_map[col] = 'Total_Fee'
                            elif 'tuition' in cl:
                                rename_map[col] = 'Tuition_Fee'
                        
                        if rename_map:
                            df.rename(columns=rename_map, inplace=True)

                        # Clean Numeric Data
                        for col in df.columns:
                            if 'fee' in col.lower() or 'amount' in col.lower():
                                df[col] = df[col].apply(clean_currency)
                        
                        # Clean Text Data
                        if 'Program' in df.columns:
                            df['Program'] = df['Program'].astype(str).str.replace('> SEE MORE', '').str.strip()

                        # Drop empty rows
                        df.dropna(how='all', inplace=True)
                        
                        all_fees.extend(df.to_dict(orient='records'))
                    else:
                        print(f"  [Table {i}] Skipped (Columns didn't match keywords)")

                except Exception as e:
                    print(f"  [Table {i}] Error processing: {e}")
                    continue

        except Exception as e:
            print(f"Error scraping {url}: {e}")
            
    return pd.DataFrame(all_fees)

# --- FUNCTION 2: Clean Policy Scraper (Unchanged/Verified) ---
def scrape_policies_clean(url_map):
    policy_data = []
    
    for policy_type, url in url_map.items():
        print(f"Scraping Policy: {policy_type}...")
        try:
            response = requests.get(url, headers=headers)
            soup = BeautifulSoup(response.content, 'html.parser')
            
            # Remove junk
            for junk in soup(['nav', 'header', 'footer', 'script', 'style', 'aside', 'form', 'noscript']):
                junk.decompose()
                
            content_div = soup.find('div', class_='entry-content') or \
                          soup.find('article') or \
                          soup.find('div', class_='col-md-9') or \
                          soup.body

            current_section = "General Introduction"
            current_text = []

            for element in content_div.find_all(['h1', 'h2', 'h3', 'h4', 'h5', 'p', 'li']):
                text = element.get_text(separator=' ', strip=True)
                if not text: continue
                
                # Heuristic: skip short menu-like items
                if len(text) < 50 and any(w in text.lower() for w in ['home', 'apply', 'contact']):
                    if element.name not in ['h1','h2','h3','h4','h5']: continue

                if element.name in ['h1', 'h2', 'h3', 'h4', 'h5']:
                    if current_text:
                        full_blob = " ".join(current_text)
                        if len(full_blob) > 20:
                            policy_data.append({
                                "Policy_Type": policy_type,
                                "Section": current_section,
                                "Content": full_blob,
                                "Source_URL": url
                            })
                    current_section = text
                    current_text = []
                else:
                    current_text.append(text)

            if current_text:
                full_blob = " ".join(current_text)
                if len(full_blob) > 20:
                    policy_data.append({
                        "Policy_Type": policy_type,
                        "Section": current_section,
                        "Content": full_blob,
                        "Source_URL": url
                    })

        except Exception as e:
            print(f"Error scraping {url}: {e}")

    return pd.DataFrame(policy_data)

# --- EXECUTION ---
print("--- STARTING SCRAPE ---")

# 1. Generate Fee Dataset
df_fees = scrape_fees_v2(fee_urls)
if not df_fees.empty:
    df_fees.to_csv("tuf_fee_structure_ready.csv", index=False)
    print(f"\nSUCCESS: Generated 'tuf_fee_structure_ready.csv' with {len(df_fees)} rows.")
else:
    print("\nWARNING: No fee data was extracted. Check the 'Accepted/Skipped' logs above.")

# 2. Generate Policy Dataset
df_policies = scrape_policies_clean(policy_urls)
if not df_policies.empty:
    df_policies.to_csv("tuf_policies_ready.csv", index=False)
    print(f"SUCCESS: Generated 'tuf_policies_ready.csv' with {len(df_policies)} sections.")
else:
    print("WARNING: No policy text found.")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import re

# --- 1. DATA INPUT (Organized by Faculty) ---
department_urls = {
    "Faculty of Medicine & Allied Health Sciences": [
        "https://tuf.edu.pk/n/departments-details/department-of-rehabilitation-sciences",
        "https://tuf.edu.pk/n/departments-details/department-of-optometry",
        "https://tuf.edu.pk/n/departments-details/department-of-nutrition-and-dietetics",
        "https://tuf.edu.pk/n/departments-details/department-of-nursing",
        "https://tuf.edu.pk/n/departments-details/department-of-professional-health-technologies",
        "https://tuf.edu.pk/n/departments-details/department-of-physiology",
        "https://tuf.edu.pk/n/departments-details/department-of-dentistry",
        "https://tuf.edu.pk/n/departments-details/department-of-medicine",
        "https://tuf.edu.pk/n/departments-details/department-of-radiology-and-medical-imaging",
        "https://tuf.edu.pk/n/departments-details/department-of-pathology",
        "https://tuf.edu.pk/n/departments-details/department-of-anesthesia"
    ],
    "Faculty of Engineering & Technology": [
        "https://tuf.edu.pk/n/departments-details/department-of-electrical-engineering",
        "https://tuf.edu.pk/n/departments-details/department-of-civil-engineering",
        "https://tuf.edu.pk/n/departments-details/department-of-chemical-engineering",
        "https://tuf.edu.pk/n/departments-details/department-of-engineering-technology"
    ],
    "Faculty of Information Technology": [
        "https://tuf.edu.pk/n/departments-details/department-of-computational-sciences"
    ],
    "Faculty of Sciences": [
        "https://tuf.edu.pk/n/departments-details/department-of-physics",
        "https://tuf.edu.pk/n/departments-details/department-of-mathematics",
        "https://tuf.edu.pk/n/departments-details/department-of-biochemistry",
        "https://tuf.edu.pk/n/departments-details/department-of-biotechnology"
    ],
    "Faculty of Management Studies": [
        "https://tuf.edu.pk/n/departments-details/national-business-school"
    ],
    "Faculty of Law": [
        "https://tuf.edu.pk/n/departments-details/department-of-law"
    ],
    "Faculty of Arts & Social Sciences": [
        "https://tuf.edu.pk/n/departments-details/department-of-design-and-interiors",
        "https://tuf.edu.pk/n/departments-details/department-of-english-language-and-literature",
        "https://tuf.edu.pk/n/departments-details/department-of-education",
        "https://tuf.edu.pk/n/departments-details/arabic-&-islamic-studies"
    ],
    "Faculty of Pharmaceutical Sciences": [
        "https://tuf.edu.pk/n/departments-details/department-of-pharmacy"
    ],
    "General Info & Student Life": [
        "https://tuf.edu.pk/n/student-societies",
        "https://tuf.edu.pk/n/international-student"
    ]
}

# --- 2. EXTRACTION LOGIC ---
def get_section_text(soup, header_keywords):
    """
    Helper function to find a header containing specific keywords 
    and extract the text immediately following it.
    """
    # Find headers (h1-h6) that contain the keyword (case insensitive)
    headers = soup.find_all(['h1', 'h2', 'h3', 'h4', 'h5', 'h6', 'strong'])
    
    for header in headers:
        if any(keyword.lower() in header.get_text().lower() for keyword in header_keywords):
            # Found the header, now grab the next sibling elements until the next header
            content = []
            curr = header.next_sibling
            while curr:
                if curr.name in ['h1', 'h2', 'h3', 'h4', 'h5', 'h6']:
                    break # Stop if we hit the next section
                if hasattr(curr, 'get_text'):
                    text = curr.get_text(strip=True)
                    if text:
                        content.append(text)
                curr = curr.next_sibling
            
            return " ".join(content)
    return ""

def scrape_department(faculty, url):
    print(f"Scraping: {url}")
    try:
        response = requests.get(url, headers={'User-Agent': 'Mozilla/5.0'})
        soup = BeautifulSoup(response.content, 'html.parser')
        
        # 1. Department Name (Usually the main H1)
        dept_name = soup.find('h1').get_text(strip=True) if soup.find('h1') else "Unknown"
        
        # 2. Main Description (All paragraph text in the main content area)
        # TUF main content is often in col-md-9
        main_content = soup.find('div', class_='col-md-9') or soup.body
        full_text = " ".join([p.get_text(strip=True) for p in main_content.find_all('p')])
        
        # 3. Vision Statement (Specific Extraction)
        vision = get_section_text(main_content, ['Vision'])
        
        # 4. Mission Statement (Specific Extraction)
        mission = get_section_text(main_content, ['Mission', 'Objective'])
        
        # 5. Programs (Heuristic: Look for lists containing 'BS', 'MS', 'PhD')
        programs = []
        all_lists = main_content.find_all('li')
        for li in all_lists:
            text = li.get_text(strip=True)
            # Check for degree keywords
            if any(x in text for x in ['BS ', 'MS ', 'MPhil', 'PhD', 'Doctor of']):
                programs.append(text)
        
        programs_str = ", ".join(list(set(programs))) # Remove duplicates
        
        return {
            "Faculty": faculty,
            "Department": dept_name,
            "Vision": vision,
            "Mission": mission,
            "Programs_Detected": programs_str,
            "Full_Description": full_text, # Good for general embedding
            "Source_URL": url
        }

    except Exception as e:
        print(f"Failed {url}: {e}")
        return None

# --- 3. MAIN EXECUTION ---
dataset = []

for faculty, urls in department_urls.items():
    for url in urls:
        data = scrape_department(faculty, url)
        if data:
            dataset.append(data)
        time.sleep(1)

# --- 4. SAVE ---
if dataset:
    df = pd.DataFrame(dataset)
    filename = "tuf_departments_details.csv"
    df.to_csv(filename, index=False)
    print(f"Saved {len(df)} departments to {filename}")
    print(df[['Department', 'Vision']].head()) # Preview
```

```python
import pandas as pd
import json
import ast
import os

# --- CONFIGURATION ---
OUTPUT_FILE = "tuf_master_dataset.jsonl"

def clean_text(text):
    if pd.isna(text): return ""
    return str(text).replace('\n', ' ').replace('\r', '').strip()

combined_data = []

# ==========================================
# 1. PROCESS COURSES
# ==========================================
try:
    # Assuming tuf_university_dataset.csv exists
    if os.path.exists("tuf_university_dataset.csv"):
        print("Processing Courses...")
        df_courses = pd.read_csv("tuf_university_dataset.csv")
        for _, row in df_courses.iterrows():
            title = clean_text(row.get('Course_Title', ''))
            if not title or title.lower() == 'nan': continue
            
            text_segment = (
                f"In the {row.get('Program_Name', 'Unknown')} program, during {row.get('Semester', 'Unknown')}, "
                f"there is a course titled '{title}' with code {row.get('Course_Code', 'N/A')}. "
                f"It is a {row.get('Category', 'General')} course worth {row.get('Credit_Hours', 'N/A')} credit hours. "
                f"Prerequisite: {row.get('Pre_Requisite', 'None')}."
            )
            combined_data.append({
                "type": "course", 
                "text": text_segment, 
                "source": row.get('Source_URL', '')
            })
    else:
        print("Skipping Courses: 'tuf_university_dataset.csv' not found.")
except Exception as e:
    print(f"Error processing courses: {e}")

# ==========================================
# 2. PROCESS FEES (The Fix)
# ==========================================
print("Processing Fees...")
try:
    df_fees = pd.read_csv("tuf_fee_structure.csv")
    
    # The CSV has headers like "('Discipline', 'Faculty of ...')"
    # We need to map these columns to their respective faculties
    faculty_map = {}
    
    # 1. Parse headers to find faculties and their specific columns
    for col in df_fees.columns:
        try:
            # Check if column is a tuple-string representation
            if col.startswith("('") and col.endswith(")"):
                col_tuple = ast.literal_eval(col)
                if isinstance(col_tuple, tuple) and len(col_tuple) == 2:
                    attr, faculty = col_tuple
                    if faculty not in faculty_map:
                        faculty_map[faculty] = {}
                    faculty_map[faculty][attr] = col
        except:
            continue

    # 2. Iterate through each faculty's data
    count_fees = 0
    for faculty, cols in faculty_map.items():
        # We need at least a Discipline (Program Name) to create a valid entry
        discipline_col = cols.get('Discipline')
        tuition_col = cols.get('Tuition Fee Per semester')
        
        if not discipline_col: continue

        # Iterate rows for this specific faculty
        for _, row in df_fees.iterrows():
            program = clean_text(row[discipline_col])
            
            # Skip if this faculty has no program in this row (side-by-side data)
            if not program or program.lower() == 'nan': continue
            
            tuition = clean_text(row.get(tuition_col, 'N/A'))
            enrolment = clean_text(row.get(cols.get('Enrolment Fee Per semester'), 'N/A'))
            total_fee = clean_text(row.get(cols.get('Total Fee Per Semester'), 'N/A'))
            
            # Get common columns (Fee Category is shared across the row)
            category = clean_text(row.get('Fee_Category', 'General'))
            source = row.get('Source_URL', '')

            text_segment = (
                f"The fee structure for the {program} program (Faculty: {faculty}) "
                f"under the category '{category}' is as follows: "
                f"Tuition Fee per semester: {tuition}. "
                f"Enrolment Fee: {enrolment}. "
                f"Total Fee per semester: {total_fee}."
            )
            combined_data.append({"type": "fee", "text": text_segment, "source": source})
            count_fees += 1
            
    print(f"   -> Added {count_fees} fee entries.")

except Exception as e:
    print(f"Error processing fees: {e}")

# ==========================================
# 3. PROCESS POLICIES
# ==========================================
print("Processing Policies...")
try:
    # Handle potentially different filenames or empty files
    policy_file = "tuf_financial_policies.csv"
    if not os.path.exists(policy_file):
        # Fallback if user has the duplicate file
        if os.path.exists("tuf_financial_policies (1).csv"):
            policy_file = "tuf_financial_policies (1).csv"
    
    if os.path.exists(policy_file):
        df_policies = pd.read_csv(policy_file)
        count_pol = 0
        for _, row in df_policies.iterrows():
            content = clean_text(row.get('Content', ''))
            if not content or content.lower() == 'nan': continue
            
            policy_type = clean_text(row.get('Policy_Type', 'Policy'))
            section = clean_text(row.get('Section_Title', 'General'))
            
            text_segment = (
                f"Regarding {policy_type}, specifically the section '{section}': {content}"
            )
            combined_data.append({
                "type": "policy", 
                "text": text_segment, 
                "source": row.get('Source_URL', '')
            })
            count_pol += 1
        print(f"   -> Added {count_pol} policy entries.")
    else:
        print("Skipping Policies: File not found.")

except Exception as e:
    print(f"Error processing policies: {e}")

# ==========================================
# 4. PROCESS DEPARTMENTS
# ==========================================
print("Processing Departments...")
try:
    if os.path.exists("tuf_departments_details.csv"):
        df_depts = pd.read_csv("tuf_departments_details.csv")
        for _, row in df_depts.iterrows():
            dept = clean_text(row.get('Department', ''))
            if not dept: continue
            
            text_segment = (
                f"The {dept} is part of the {row.get('Faculty', 'University')}. "
                f"Vision: {clean_text(row.get('Vision', 'N/A'))} "
                f"Mission: {clean_text(row.get('Mission', 'N/A'))} "
                f"Programs offered include: {clean_text(row.get('Programs_Detected', 'N/A'))}."
            )
            combined_data.append({
                "type": "department", 
                "text": text_segment, 
                "source": row.get('Source_URL', '')
            })
    else:
        print("Skipping Departments: File not found.")
except Exception as e:
    print(f"Error processing departments: {e}")

# ==========================================
# SAVE OUTPUT
# ==========================================
print(f"Saving {len(combined_data)} documents to {OUTPUT_FILE}...")
with open(OUTPUT_FILE, 'w', encoding='utf-8') as f:
    for entry in combined_data:
        json.dump(entry, f)
        f.write('\n')

print("Done!")
```
